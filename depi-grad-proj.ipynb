```python
import torch
from langchain_community.embeddings import HuggingFaceEmbeddings 
from transformers import AutoTokenizer, AutoModelForCausalLM, BitsAndBytesConfig, pipeline

# --- LLM ---
embedding_model_id = "intfloat/multilingual-e5-large"
llm_model_id = "Qwen/Qwen2.5-7B-Instruct"

def load_models():
    print(f"⏳ Loading Embeddings: {embedding_model_id}...")
    embeddings = HuggingFaceEmbeddings(
        model_name=embedding_model_id,
        model_kwargs={'device': 'cuda'}
    )
    
    print(f"⏳ Loading LLM: {llm_model_id}...")
    # 4-bit Quantization 
    bnb_config = BitsAndBytesConfig(
        load_in_4bit=True,
        bnb_4bit_compute_dtype=torch.float16,
        bnb_4bit_quant_type="nf4"
    )
    
    tokenizer = AutoTokenizer.from_pretrained(llm_model_id)
    model = AutoModelForCausalLM.from_pretrained(
        llm_model_id,
        quantization_config=bnb_config,
        device_map="auto"
    )
    
    pipe = pipeline(
        "text-generation",
        model=model,
        tokenizer=tokenizer,
        max_new_tokens=1024,
        temperature=0.3, 
        repetition_penalty=1.1,
        return_full_text=False
    )
    
    print("✅ Models Loaded Successfully!")
    return embeddings, pipe

embeddings, llm_pipe = load_models()
```

```python
from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter 
from langchain_community.vectorstores import FAISS


pdf_path = "/kaggle/input/egyptian-constitution/Egyptian Constitution.pdf"

def create_vector_db(pdf_path, _embeddings):
    print("⏳ Loading PDF...")
    try:
        loader = PyPDFLoader(pdf_path)
        docs = loader.load()
        
        # Split text into chunks
        text_splitter = RecursiveCharacterTextSplitter(
            chunk_size=800, 
            chunk_overlap=200,
            separators=["\n\n", "\n", ".", " "]
        )
        splits = text_splitter.split_documents(docs)
        print(f"📄 Document split into {len(splits)} chunks.")
        
        # Create Vector Store
        print("⏳ Indexing data (this may take a minute)...")
        vectorstore = FAISS.from_documents(splits, _embeddings)
        print("✅ Vector Database Ready!")
        return vectorstore
        
    except Exception as e:
        print(f"❌ Error: {e}")
        return None

vector_db = create_vector_db(pdf_path, embeddings)
```

```python
def ask_constitution(query, vector_db, llm_pipe):
    # 1. Retrieve Context
    # get 5 chunks to ensure we catch related articles
    retriever = vector_db.as_retriever(search_kwargs={"k": 5})
    docs = retriever.invoke(query)
    
    # Combine retrieved text
    context_text = "\n\n".join([f"[Article Part {i+1}]: {d.page_content}" for i, d in enumerate(docs)])
    
    # 2. Construct the Prompt (Arabic)
    # This prompt forces the model to act as a legal consultant
    system_prompt = """أنت مستشار قانوني خبير في الدستور المصري.
    مهمتك هي الإجابة على سؤال المستخدم بناءً *فقط* على النصوص الدستورية المقدمة أدناه.
    
    التعليمات:
    1. اقرأ النصوص المقدمة بعناية.
    2. لا تستخدم معلومات خارجية إذا لم تكن موجودة في النصوص.
    3. اشرح إجابتك بوضوح واستشهد برقم المادة إن وجد.
    4. إذا كانت الإجابة غير موجودة في النصوص، قل "لا توجد معلومات كافية في الوثائق المتاحة".
    """
    
    user_message = f"""
    النصوص الدستورية المتاحة (Context):
    {context_text}
    
    سؤال المستخدم:
    {query}
    """
    
    messages = [
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": user_message}
    ]
    
    result = llm_pipe(messages)
    return result[0]['generated_text'], context_text
```

```python
# Test Query
query = "هل يجوز محاكمة المدنيين أمام القضاء العسكري؟" 

print(f"❓ Question: {query}\n")

response, sources = ask_constitution(query, vector_db, llm_pipe)

print("-" * 50)
print("🤖 AI Answer:")
print(response)
print("-" * 50)
print("📚 Sources Used:")
print(sources)
```
